# 02 — YOLOP Demo (Lane + Drivable Area + Vehicles)

Quick test of YOLOP inference on sample images.
YOLOP handles lane detection, drivable area segmentation, and vehicle detection in one forward pass.

```python
import sys
sys.path.insert(0, '../src')

import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from utils import PROJECT_ROOT, MODELS_DIR
```

```python
# Option A: Load via ONNX
from yolop_inference import YOLOPInference

onnx_path = MODELS_DIR / 'yolop.onnx'
if onnx_path.exists():
    yolop = YOLOPInference(str(onnx_path), backend='onnx')
else:
    # Option B: PyTorch Hub (downloads model automatically)
    yolop = YOLOPInference(backend='pytorch')
```

```python
# Test on a sample image
# Replace with your own dashcam image
img_path = '../data/sample_road.jpg'  # Add a road image here

if Path(img_path).exists():
    frame = cv2.imread(img_path)
    results = yolop.infer(frame)
    output = yolop.draw_results(frame, results)
    
    plt.figure(figsize=(14, 8))
    plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
    plt.title(f'YOLOP: {len(results["det_boxes"])} vehicles, '
              f'lane pixels: {results["lane_seg"].sum()}')
    plt.axis('off')
    plt.show()
else:
    print(f'Place a road image at {img_path} to test')
```
